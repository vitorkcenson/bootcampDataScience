# Logistic Regression Optional Task

Develop a classifier that can determine whether a given iris flower including all three categories, Iris-setosa, Iris-versicolour, and Iris-virginica corresponding to the numeric values 0, 1, and 2, respectively. This will now be a multiclass problem. Observe how this changes the confusion matrix.

```python
# Importing necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn import preprocessing
```

```python
# Loading the Iris dataset
df_iris = pd.read_csv("Iris.csv")
df_iris.head()
```

```python
# Create a dictionary mapping species names to numerical values
species_mapping = {'Iris-setosa': 0, 'Iris-versicolor': 1, 'Iris-virginica': 2}

# Use the map function to encode the 'species' column
df_iris['species_codes'] = df_iris['Species'].map(species_mapping)
```

```python
df_iris.tail()
```

```python
# Identify your independent variable x
# Select feature columns
X = df_iris.iloc[:,[1,2,3,4]].values

# Select the target column ('Species')
y = df_iris.iloc[:,6].values

# Reshape X to have 4 features per row
X = X.reshape(-1, 4)
# Scale the feature values for easier model fitting
X = preprocessing.scale(X)
```

```python
print(y)
```

```python
# Split the data into training and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)
```

```python
# Fit the logistic regression model to the training data
log_reg = LogisticRegression()
log_reg.fit(X_train, y_train)

# Make predictions on the test data and reshape the output
y_pred = log_reg.predict(X_test).reshape(-1,1)
```

```python
from sklearn.metrics import confusion_matrix

# Create a list of class codes from the Iris dataset
classes = list(pd.Series(y_test).unique())

# Generate the confusion matrix using the true labels (y_test) and predicted labels (y_pred)
conf_mat = confusion_matrix(y_test, y_pred)

print(conf_mat)
```

```python
# Convert the confusion matrix into a DataFrame for better visual representation
cm_df = pd.DataFrame(conf_mat, columns=classes, index=classes)
cm_df
```

0 corresponds to the Iris-setosa class.
1 corresponds to the Iris-versicolour class.
2 corresponds to the Iris-virginica class.

```python
from sklearn.metrics import f1_score, precision_score, recall_score

# Calculate the average F1 score
av_f1 = f1_score(y_test, y_pred, average='micro')
print('F1 Score:', av_f1)

# Calculate the F1 score for each individual class
f = f1_score(y_test, y_pred, average=None)

# Identify the lowest F1 score among the calculated scores
lowest_score = min(f)

# Determine the class corresponding to the lowest F1 score
difficult_class = classes[list(f).index(lowest_score)]

# Print the most challenging class based on the lowest F1 score
print('Most challenging class:', difficult_class)
```

```python
# Calculate accuracy, precision and recall manually

# Remember that conf_mat = confusion_matrix(y_test, y_pred)
# And classes = list(pd.Series(y_test).unique())  
# Get unique class labels (important!)

num_classes = conf_mat.shape[0]  # Number of classes

precision_per_class = []
recall_per_class = []
accuracy_per_class = [] # Added accuracy per class

for i in range(num_classes):
    tp = conf_mat[i, i]
    fp = np.sum(conf_mat[:, i]) - tp
    fn = np.sum(conf_mat[i, :]) - tp

    accuracy = tp  / (tp + fp + fn) if (tp + fp + fn) > 0 else 0.0 # Calculate per class accuracy                                                                 
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0

    precision_per_class.append(precision)
    recall_per_class.append(recall)
    accuracy_per_class.append(accuracy)

print("Confusion Matrix:")
print(conf_mat)
print("Classes:", classes)

for i, class_name in enumerate(classes):
    # Print per class accuracy
    print(f"\nMetrics for class {class_name}:")
    print(f"  Precision: {precision_per_class[i]}")
    print(f"  Recall: {recall_per_class[i]}")
    print(f"  Accuracy: {accuracy_per_class[i]}")
```
